```python
import pandas as pd
```

```python
birth_rate = pd.read_csv(r"C:/Downloads/new_data/birth_rate.csv")
eu_regions = pd.read_csv(r"C:/Downloads/new_data/eu_country.csv")
birth_rate.head()
```

```python
birth_rate.columns
```

```python
# kiểm tra các giá trị có thể có
birth_rate["MEASURE"].unique()
```

```python
# chọn các cột quan trọng, đổi tên
test = birth_rate[["TIME_PERIOD","REF_AREA","Reference area","Age","OBS_VALUE"]]
test = test.rename(columns= { "Reference area" : "Country",
                      "TIME_PERIOD" : "Year",
                      "OBS_VALUE" : "FERT_RATIO"

})
```

```python
#phân loại và lọc các giá trị phù hợp
age = test["Age"] == "Total"
test = test[age]
test["FERT_RATIO"] = test["FERT_RATIO"].dropna()
test
```

```python
# lọc các nước nằm trong châu âu
eu = test["Country"].isin(eu_regions["Country"])
test = test[eu]

#reset lại chỉ mục
test =test.reset_index(drop = True)
```
