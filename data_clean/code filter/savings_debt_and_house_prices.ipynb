```python
import pandas as pd
```

```python
credits = pd.read_csv(r"C:/Downloads/new_data/tiết kiệm và nợ của hộ gia đình.csv")
eu_regions = pd.read_csv(r"C:/Downloads/new_data/eu_country.csv")
credits.head()
```

```python
credits.columns
```

```python
cols_keep = [
    "TIME_PERIOD",
    "REF_AREA",
    "Zone de référence",
    "Mesure",
    "UNIT_MEASURE",
    "OBS_VALUE",
]
```

```python
test = credits[cols_keep]
test
```

```python
rename_map = {
    "Zone de référence": "Country",
    "TIME_PERIOD": "Year",
    "Mesure": "Measure",
    "UNIT_MEASURE": "Unit_measure",
    "OBS_VALUE": "Value"
}

test = test.rename(columns=rename_map)
test
```

```python
eu_country = test["REF_AREA"].isin(eu_regions["REF_AREA"])
test = test[eu_country]
test
```

```python
test["Measure"].unique()

```

```python
measure_map = {
    "Dette des ménages et des ISBLSM": "household_debt",
    "Épargne nette des ménages et des ISBLSM": "household_net_saving",
    "Terrains par tête des ménages et des ISBLSM": "land_per_capita"
}

test["Measure"] = test["Measure"].replace(measure_map)
test = test[test["Measure"].isin(measure_map.values())]
test
```

```python
test["Unit_measure"].unique()
```

```python
map = {"PT_B6N_S1M" : "PT", "USD_PPP_PS" : "USD_PPP"}
test["Unit_measure"] = test["Unit_measure"].replace(map)
test
```

```python
test = test.dropna(subset = "Value")
```

```python
test["Value"] = test["Value"].round(2) 
test
```

```python
measure_sets = (
    test
    .groupby(["Year", "Country"])["Measure"]
    .apply(set)
    .reset_index(name="measures")
)

measure_sets[measure_sets["measures"].apply(len) < 3]
```
